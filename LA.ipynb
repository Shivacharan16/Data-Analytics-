## Matrix Operations

```python
# Matrix addition nad subtraction
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A-B)
print(A+B)
```

```python
scaler=10
A=np.array([[1,2],
            [3,4]])
print(scaler*A)
```

```python
# Dot products on vectors
a=np.array([1,2])
b=np.array([3,4])
print(a@b)
print(np.dot(a,b))
```

```python
# Dot products on matrix
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A@B)
```

```python
# Transpose of matrix ( row to col and col to row)
A=np.array([[1,2],
            [3,4]])
print(np.transpose(A))
print(A.T)
```

```python
# Determinant of matrix 
A=np.array([[1,2],
            [3,4]])
print(np.linalg.det(A))
```

```python
# Inverse of matrix
A=np.array([[1,2],
            [3,4]])
print(np.linalg.inv(A))
```

# Measure of dispersion

```python
# Spreadness of data : how much the data is far away from mean /median
# MAD ,Variance,Strandard devation ,percentile,Quantile, IQR, five number summary
```

```python
import numpy as np
# mean absolute deviation(MAD)

reelcount = [50,10,15,20,50,10,10,11]

mean = np.mean(reelcount)
print(mean)

int_result = []

for i in reelcount:
    int_result.append(np.abs(i-mean))

print(int_result)

MAD = (1/len(reelcount))*np.sum(int_result)
print(MAD)
```

```python
# Variance
reelcount = [50,10,15,20,50,10,10,11]

mean = np.mean(reelcount)

int_result = []

for i in reelcount:
    int_result.append((np.abs(i-mean))**2)

variance = (1/len(reelcount))*np.sum(int_result)
print(variance)
```

```python
var = np.var(reelcount)
print(var)
```

```python
# stranderd devation
print(np.std(reelcount))
```

```python
SD = np.sqrt(var)
print(SD)
```

```python
# percentile
reelcount = [50,10,15,20,50,70,80,11,5,2]
Q1 = np.percentile(reelcount,25)
Q2 = np.percentile(reelcount,50)
Q3 = np.percentile(reelcount,75)
IQR =Q3-Q1
LB = Q1-1.5*IQR
UB = Q3+1.5*IQR
print(Q1,Q3,IQR,LB,UB)
```

#   2.1.1 sample sampling techniques

```python
# create the population data

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)
population  = pd.DataFrame({
                'student_id':np.arange(1,10001),
                'canteen_spending' : np.random.gamma(10,50,10000),
                'food_items' : np.random.choice(['puff','samosa','biryani','cool drinks'],10000),
                'gender' : np.random.choice(['girls','boys'],10000,p=[0.4,0.6])
})
population
```

```python
# sample random sampling take the random samples from the population.
# close the eyes and  taking 500 random sample from 10000 records
simple_sample = population.sample(n = 500,random_state = 42)
simple_sample.describe()
```

# systematic sampling

```python
# pick every kth element in population as sample
# iloc

systematic_sample = population.iloc[::20]
systematic_sample
```

# stratified sampling

```python
# if we want same ration of boys & girls in sample data such as in population data
# stratified sampling make sure category ratio of population and samples match

from sklearn.model_selection import train_test_split
stratified_sampling,_ = train_test_split(population,stratify = population['gender'],train_size = 0.05)
stratified_sampling.gender.value_counts()
```

# cluster sampling

```python
cluster_sample = population[(population.food_items == 'biryani')|(population.food_items == 'cool drinks')]
cluster_sample
```

# 2.2 Central Limit Theorem (CLT)

```python
population = np.random.exponential(scale = 2 ,size = 100000)
population = pd.Series(population)
sample_size = 50
num_sample = 1000
sample_mean = [] # list to store
for i in range (num_sample):
    sample  = population.sample(n = sample_size)
    mean_value = sample.mean()
    sample_mean.append(mean_value)

sns.histplot(sample_mean,kde  = True)
```
