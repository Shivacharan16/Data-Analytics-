## Matrix Operations

```python
# Matrix addition nad subtraction
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A-B)
print(A+B)
```

```python
scaler=10
A=np.array([[1,2],
            [3,4]])
print(scaler*A)
```

```python
# Dot products on vectors
a=np.array([1,2])
b=np.array([3,4])
print(a@b)
print(np.dot(a,b))
```

```python
# Dot products on matrix
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A@B)
```

```python
# Transpose of matrix ( row to col and col to row)
A=np.array([[1,2],
            [3,4]])
print(np.transpose(A))
print(A.T)
```

```python
# Determinant of matrix 
A=np.array([[1,2],
            [3,4]])
print(np.linalg.det(A))
```

```python
# Inverse of matrix
A=np.array([[1,2],
            [3,4]])
print(np.linalg.inv(A))
```

# Measure of dispersion

```python
# Spreadness of data : how much the data is far away from mean /median
# MAD ,Variance,Strandard devation ,percentile,Quantile, IQR, five number summary
```

```python
import numpy as np
# mean absolute deviation(MAD)

reelcount = [50,10,15,20,50,10,10,11]

mean = np.mean(reelcount)
print(mean)

int_result = []

for i in reelcount:
    int_result.append(np.abs(i-mean))

print(int_result)

MAD = (1/len(reelcount))*np.sum(int_result)
print(MAD)
```

```python
# Variance
reelcount = [50,10,15,20,50,10,10,11]

mean = np.mean(reelcount)

int_result = []

for i in reelcount:
    int_result.append((np.abs(i-mean))**2)

variance = (1/len(reelcount))*np.sum(int_result)
print(variance)
```

```python
var = np.var(reelcount)
print(var)
```

```python
# stranderd devation
print(np.std(reelcount))
```

```python
SD = np.sqrt(var)
print(SD)
```

```python
# percentile
reelcount = [50,10,15,20,50,70,80,11,5,2]
Q1 = np.percentile(reelcount,25)
Q2 = np.percentile(reelcount,50)
Q3 = np.percentile(reelcount,75)
IQR =Q3-Q1
LB = Q1-1.5*IQR
UB = Q3+1.5*IQR
print(Q1,Q3,IQR,LB,UB)
```

# Covariance

```python
height = [170,150,140]
weight = [80,70,60]
cov_matrix = np.cov(height,weight)
print(cov_matrix)
```

```python
height = [170,150,140]
weight = [60,70,80]
# sample covariance 
cov_matrix = np.cov(height,weight)
print(cov_matrix)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

vec1 = [[1,2,3]]
vec2 = [[4,5,6]]

similarity = cosine_similarity(vec1,vec2)
print(similarity)
```

```python
# cosine  distance  =  1 - cosine similarity
from sklearn.metrics.pairwise import cosine_similarity,cosine_distances

vec1 = [[1,2,3]]
vec2 = [[4,5,6]]

similarity = cosine_similarity(vec1,vec2)
distance  = cosine_distances(vec1,vec2)

print(similarity)
print(distance)
```

```python
# cosine  distance  =  1 - cosine similarity
print(1-similarity)
```
