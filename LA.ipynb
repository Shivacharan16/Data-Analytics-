# Linear Algebra

### Scaler, Vector, Matrix and Tensor

```python
import numpy as np
```

```python
# Scaler : 0D array
salary=np.array(500000)
print(salary)
print(salary.ndim)
print(type(salary))
```

```python
# vector:1D array
expense=np.array([6000,4000,10000000,2000])
print(expense)
print(expense.ndim)
print(type(expense))
print(expense.shape)
```

```python
# Matrix:2D array
dataframe=np.array([[6000,4000,10000,2000],
                    [20000,400,29999,40000]])
print(dataframe)
print(dataframe.ndim)
print(type(dataframe))
print(dataframe.shape)
```

```python
# Tensor:3D or more D arry
array=np.array([[[1,2,3,4,5],
                 [2,3,1,3,4]],
               [[4,8,5,3,6],
               [2,5,3,3,2]]])
print(array)
print(array.ndim)
print(type(array))
print(array.shape)
```

```python
# 4D array
array=np.array([[[[1,2,3,4,5],
                 [2,3,1,3,4]],
                [[4,8,5,3,6],
                 [2,5,3,3,2]]],
                [[[1,2,3,4,5],
                 [2,3,1,3,4]],
                [[4,8,5,3,6],
                 [2,5,3,3,2]]]])
print(array)
print(array.ndim)
print(type(array))
print(array.shape)
```

## Matrix Operations

```python
# Matrix addition nad subtraction
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A-B)
print(A+B)
```

```python
scaler=10
A=np.array([[1,2],
            [3,4]])
print(scaler*A)
```

```python
# Dot products on vectors
a=np.array([1,2])
b=np.array([3,4])
print(a@b)
print(np.dot(a,b))
```

```python
# Dot products on matrix
A=np.array([[1,2],
            [3,4]])
B=np.array([[4,7],
            [6,8]])
print(A@B)
```

```python
# Transpose of matrix ( row to col and col to row)
A=np.array([[1,2],
            [3,4]])
print(np.transpose(A))
print(A.T)
```

```python
# Determinant of matrix 
A=np.array([[1,2],
            [3,4]])
print(np.linalg.det(A))
```

```python
# Inverse of matrix
A=np.array([[1,2],
            [3,4]])
print(np.linalg.inv(A))
```
