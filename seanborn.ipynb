```python
# def seanborn is high level visualization library built on the top of matplotlib.
# seanborn library is popularly used for plotting statistical charts e.g box plot,KDE plot, pair plot.
# seanborn provides the inbulid datasets also
# matplotlib provideb detailed control on the chart while seaborn don't provide  detailed 
```

```python
import seaborn as sns
```

```python
# datasets inside the section
print(sns.get_dataset_names())
```

```python
# titanic dataset
df = sns.load_dataset("titanic")
df
```

```python
df = sns.load_dataset("flights")
df
```

# line chart
# all the charts of matplotlib can also be plotted by seaborn

```python
# trends over thetime
# on the same chart features of seanborn and mathplotlib can be apply
import matplotlib.pyplot as plt
timeline = df.groupby("year")['passengers'].sum().reset_index()

sns.lineplot(x ="year",y="passengers",data=timeline,marker = 'o')
plt.title("year wise passengers")
plt.grid()
```

```python
df=sns.load_dataset('tips')
df
```

```python
# to find the relation between two varaibles
plt.figure(figsize = (4,3))
sns.scatterplot(x = 'total_bill',y = 'tip',data = df,hue= 'smoker')
```

```python
# pair plot : compare all the possible numerical pair of varaibles at a time 
# compare multiple columns with each other together
plt.figure(figsize=(3,2))
sns.pairplot(df,hue="smoker",kind='scatter')
```

# heatmap

```python
# USE: heatmap visualize the exactc correalation b/w numerical/catigorical variables
# heatmap is used for ploting correction matrix in colorful chart
import pandas as pd
# df.corr() function is used generate correlation table
corr = df.corr(numeric_only = True)
print(corr)
```

```python
sns.heatmap(corr,annot =True,cmap ='BrBG',fmt ='.3f')
```

```python
df = sns.load_dataset("car_crashes")
df
```

```python
plt.figure(figsize = (4,3))
sns.scatterplot(x = 'speeding',y = 'alcohol',data = df)
```

```python
sns.pairplot(df,kind = 'scatter')
```

```python
co = df.corr(numeric_only=True)
sns.heatmap(co,annot =True,cmap ='Pastel1',fmt ='.2f')
```

```python
timeline = df.groupby("speeding")['alcohol'].mean().reset_index()
sns.lineplot(x ="speeding",y="alcohol",data=timeline,marker = 'o')
plt.grid()
```
