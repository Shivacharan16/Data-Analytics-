# Basic plots: Bar chart, Histogram, Line chart, Scatter plot,Bar chart

```python
# Use : to compare categorical and discreate data
```

```python
food_cat = ['Biryani','Idli','Dosa','Shawarma']
like_count = [50,10,20,30]
plt.figure(figsize=(4,3))
plt.bar(food_cat,like_count,color = ['blue','red','green','gold'])

# to mention the value over the bar
for i,count in enumerate(like_count):
    plt.text(i,count+1,str(count),ha = 'center')

# i: x coordinate
# count+1 : y-coordinate
# str(count) : value to print on the chart
# ha = 'center' : horizontally placing value at the center of the bar
        
plt.xlabel("food categories")
plt.ylabel("like count")
plt.title("food categories by like count")
```

# Histogram

```python
# Use: Histogram works on continuouns data 
# used to show the frequency of data in a particular range
# Histogram is popularly used for checking the shape of the data
```

```python
import numpy as np
data = np.random.randn(1000)  # random normal distributed data
plt.figure(figsize=(5,3))
plt.hist(data,bins=25,edgecolor = 'white',color = 'red')

plt.xlabel('continuouns data ')
plt.ylabel('frequency of data ')
plt.show()

```

```python
# Interview Question :
# what is the difference between bar and Histogram chart/
#ans :
# 1. bar chart works on categorical /discreate data while histogram works on continuouns data
# 2. bar chart is used for comparing categories/discreate data while histgram is used for checking the shape of the data
```

# line chart

```python
#  trends over the time
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
icecream_sales=[90,20,50,40,60,70,50,60,80,100,40,50]

plt.plot(dates, icecream_sales, marker = 'o')
plt.xlabel("month wise sales of icecream")
plt.ylabel("total no.of sales")
plt.xticks(dates, rotation = 45)
plt.show()
```

```python
dates = pd.date_range(start = "2026-01-01", periods = 12, freq = 'ME')
icecream_sales=[90,20,50,40,60,70,50,60,80,100,40,50]

plt.plot(dates, icecream_sales, marker = 'o')

for x,y in zip(dates, icecream_sales):
    plt.text(x,y, str(y))
plt.xticks(dates,dates.strftime("%b"), rotation = 35)
plt.grid()
plt.show()
```

### Multiple line chart

```python
# used to compare the two or more trends side by side
```

```python
dates = pd.date_range(start = "2015-01-01", periods = 12, freq = 'YE')
flipkart_sales=[90,70,10,5,25,46,90,42,53,110,120,30]
amazon_sales=[100,50,60,120,40,60,50,50,45,25,110,50]

plt.plot(dates, flipkart_sales, marker = 'o', label = "FLIPKART")
plt.plot(dates, amazon_sales, marker = '*', label = "AMAZON")

plt.legend()
plt.grid()
```

### pie chart

```python
# pie chart is used to show the categorical data 
# pie chart is to show the proportion of the whole 

# we use the pie chart when no.of categories are less
```

```python
# mobile market share with pie chart

brands = ["vivo", "mi", "apple", "samsung", "moto", "others"]
market_share = [15,10,32,30,8,5]
plt.pie(market_share, labels = brands, autopct = '%1.1f%%')

plt.savefig("pie_chart DA-22.png")
plt.show()
```

### Donut chart

```python
# it is same like pie chart but the shape almost similar to pie only
```

```python
brands = ["vivo", "mi", "apple", "samsung", "moto", "others"]
market_share = [15,10,32,30,8,5]
plt.pie(market_share, labels = brands, autopct = '%1.1f%%', wedgeprops = dict(width = 0.6))
plt.show()
```
