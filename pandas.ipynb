# 1. data structure creation (series(1D array),dataframe(2D array))

### series (1D array)

```python
# create series
ser1 =pd.Series()
print(ser1)
```

```python
# conver list to Series 
list1 = [10,20,30,40,50]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# conver list to Series 
list1 = [10,20,30,40,50,'hello']
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
# in pandas strings are represented as object
# imp: in series, all elements have  the same datatype
```

```python
# example-2: convert tuple to Series
data = (10,20,30,50)
ser3=pd.Series(data)
print(ser3)
print(type(ser3))
print(ser3[2])

# in pandas series direct negative indexing not allowed.
```

```python
# convert the numpy array into series
import numpy as np
arr1 = np.array([10,20,30,40,50])
ser4 = pd.Series(arr1,index=['a','b','c','d','e'])
print(arr1,type(arr1))
print(ser4,type(ser4))
print(ser4['a'])
```

```python
# update operation on series
ser3[3]=450
# insert new value
ser3[5] = 550
ser3['a'] =650
ser3[60] =750
ser3[1.2] = 950
```

```python
print(ser3)
```

```python
# delete element form series
ser2 = ser2.drop(1)
ser2
```

### creating data frame

### creating new column in the dataframe

```python
#  here rank() function is a pandas function
# rank() function is giving rank-1 to minmum salary and higher rank value to higher salary
df_excel['rank'] = df_excel['salary'].rank(ascending =False)
df_excel.sort_values('rank')

# sort_values() is used for sorting  the dataframe based upon a column values
```

```python
# droping the column
# axis =1 : delete the column
# axis =0 : delete the row
df_excel=df_excel.drop(['rank'],axis=1)
df_excel
```
