# 1. data structure creation (series(1D array),dataframe(2D array))

### series (1D array)

```python
# create series
ser1 =pd.Series()
print(ser1)
```

```python
# conver list to Series 
list1 = [10,20,30,40,50]
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
```

```python
# conver list to Series 
list1 = [10,20,30,40,50,'hello']
ser2 = pd.Series(list1)
print(ser2)
print(type(ser2))
# in pandas strings are represented as object
# imp: in series, all elements have  the same datatype
```

```python
# example-2: convert tuple to Series
data = (10,20,30,50)
ser3=pd.Series(data)
print(ser3)
print(type(ser3))
print(ser3[2])

# in pandas series direct negative indexing not allowed.
```

```python
# set cannot be converted to pandas series
# set is unordered and series is ordered, so no conversation
data = {10,20,30,50}
ser3=pd.Series(data)
print(ser3)
```

```python
# convert the numpy array into series
import numpy as np
arr1 = np.array([10,20,30,40,50])
ser4 = pd.Series(arr1,index=['a','b','c','d','e'])
print(arr1,type(arr1))
print(ser4,type(ser4))
print(ser4['a'])
```

```python
# update operation on series
ser3[3]=450
# insert new value
ser3[5] = 550
ser3['a'] =650
ser3[60] =750
ser3[1.2] = 950
```

```python
print(ser3)
```

```python
# delete element form series
ser2 = ser2.drop(1)
ser2
```

### creating data frame

```python
import pandas as pd
```

```python
df = pd.DataFrame()
print(df)
```

```python
# converting list to dataframe
fruits_list = ['aratipandu','mamadipandu','kamalapandu','drakshapandu']
df=pd.DataFrame(fruits_list,columns = ['fruit_name'])
df
```

```python
# converting 2D list to DataFrame
fruits_list = [['aratipandu',40],['mamadipandu',100],['kamalapandu',20],['drakshapandu',480]]
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price'])
df
```

```python
fruits_list = [('aratipandu',40),('mamadipandu',100),('kamalapandu',20),('drakshapandu',480)]
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price'])
df
```

```python
# converting 2D list to DataFrame using 3 columns
fruits_list = [['aratipandu',40,10],['mamadipandu',100,90],['kamalapandu',20,5],['drakshapandu',480,120]]
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price','dicount'])
df
```

```python
# converting 2D list to DataFrame using 3 columns
fruits_list = {('aratipandu',40,10),('mamadipandu',100,90),('kamalapandu',20,5),('drakshapandu',480,120)}
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price','dicount'])
df
```

```python
# creating DataFram from set of tuples
fruits_list = {('aratipandu',40,10),('mamadipandu',100,90),('kamalapandu',20,5),('drakshapandu',480,120)}
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price','dicount'])
df
```

```python
# creating DataFram from set of set we cannot create because set is mutable, list is also mutable
# in set we cannot store the mutable items.
# in set only immutable items can store
fruits_list = {{'aratipandu',40,10},{'mamadipandu',100,90},{'kamalapandu',20,5},{'drakshapandu',480,120}}
df=pd.DataFrame(fruits_list,columns = ['fruit_name','price','dicount'])
df
```

```python
# creating data frame from dictionary
st_data = {'name':['shiva','niranjan','harish','cherry','sai'],'salary':[40,50,60,15,25],'stu_time':[6,4,5,1,3]}
df= pd.DataFrame(st_data)
df
```
