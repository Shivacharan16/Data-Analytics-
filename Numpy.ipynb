# 2. Creating Numpy arrays

```python
# Example-1: numpy array of interger
arr1 = np.array([1,2,3,4,5])
print(arr1)
print(arr1.dtype)
type(arr1)
```

```python
# Example-2: numpy array of float
arr2 = np.array([1.1,2.5,3.8,4.2,5.6])
print(arr2)
print(arr2.dtype)
type(arr2)
```

```python
# Example-3: numpy array of float + int
arr3 = np.array([1,2.5,3.8,4,5])
print(arr3)
print(arr3.dtype)
type(arr3)
```

```python
# Example-4: numpy array of strings
arr4 = np.array(['shiva','ch@ran','hi'])
print(arr4)
print(arr4.dtype)
type(arr4)

# U : unicode : Special symbols are allowed  in unicode : strings
```

```python
# Example-5: Creating 2D numpy array
arr5 = np.array([[1,2,3],[4,5,6]])
print(arr5)
print(type(arr5))
print(arr5.dtype)
print(arr5.ndim) #ndim is used to check the dimension of array
```

```python
# Example-6: Creating 3D numpy array
arr6 = np.array([[[1,2,3],[4,5,6]],[[10,20,30],[40,50,60]]])
print(arr6)
print(type(arr6))
print(arr6.dtype)
print(arr6.ndim)
```
