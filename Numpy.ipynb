## creating sequences,shuffle,permutation and lineraly seperated data

```python
# arrange() function is used for creating a sequence of elements
# arange() is just similary to range() function
import numpy as np
arr1 = np.arange(10)
print(arr1)
```

```python
arr1 = np.arange(1,10)
print(arr1)
```

```python
arr1 = np.arange(1,10,2)
print(arr1)
```

```python
# shuffle() : used for shuffle the elements 
# shuffle() perform inplace operation 
arr1 = np.arange(10)
print(arr1)

np.random.shuffle(arr1)
arr1
```

```python
# permutation() : shuffle elements but returns new shuffled array
# this don't modify the orginal array
arr1 = np.arange(10)
print(arr1)

arr2 = np.random.permutation(arr1)
print(arr2)
print(arr1)
```

```python
# linspace(): linearly spaced array
# linspace(start,end,numbermof elements)
arr3 = np.linspace(1,10,10)
print(arr3)
```

```python
arr3 = np.linspace(1,10,4)
print(arr3)
```

```python
arr3 = np.linspace(1,10,50)
print(arr3)
```

```python
# Modification, insertion and deletion in numpy
import numpy as np
s_s = np.array([["motichoor","kalakand","jalabi"],["poli","gariyalu","khaja"]])
# Modification
s_s[0,0] = 'payasum'
print(s_s)
```

```python
# updated complete row
s_s[1] = ['kesari','pulareddy','arasalu']
print(s_s)
```

```python
# updated complete column
# numpy array take fix space to store data
s_s[:,2] = ['mysurepak','doublekamitha']
print(s_s)
```

```python
# Insert operation : new row
s_s = np.insert(s_s,1,['gajarpak','rusmali','firni'], axis = 0)
print(s_s)
```

```python
# insert operation : new column  
s_s = np.insert(s_s,2,['radbhari','gulabjamun','kheer'], axis = 1)
print(s_s)
```

```python
# insert operation : multiple columns
s_s = np.insert(s_s,2,[['kajukatli','ghevar','peda'],['kajukatli','ghevar','peda']], axis = 1)
print(s_s)
```

```python
# delete operation 
s_s=np.delete(s_s,0, axis=0)
print(s_s)
```

```python
s_s=np.delete(s_s,2, axis=1)
print(s_s)
```

```python
s_s=np.delete(s_s,[1,3], axis=1)
print(s_s)
```
