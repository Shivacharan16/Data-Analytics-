# 2. Creating Numpy arrays

```python
# Example-1: numpy array of interger
arr1 = np.array([1,2,3,4,5])
print(arr1)
print(arr1.dtype)
type(arr1)
```

```python
# Example-2: numpy array of float
arr2 = np.array([1.1,2.5,3.8,4.2,5.6])
print(arr2)
print(arr2.dtype)
type(arr2)
```

```python
# Example-3: numpy array of float + int
arr3 = np.array([1,2.5,3.8,4,5])
print(arr3)
print(arr3.dtype)
type(arr3)
```

```python
# Example-4: numpy array of strings
arr4 = np.array(['shiva','ch@ran','hi'])
print(arr4)
print(arr4.dtype)
type(arr4)

# U : unicode : Special symbols are allowed  in unicode : strings
```

```python
# Example-5: Creating 2D numpy array
arr5 = np.array([[1,2,3],[4,5,6]])
print(arr5)
print(type(arr5))
print(arr5.dtype)
print(arr5.ndim) #ndim is used to check the dimension of array
```

```python
# Example-6: Creating 3D numpy array
arr6 = np.array([[[1,2,3],[4,5,6]],[[10,20,30],[40,50,60]]])
print(arr6)
print(type(arr6))
print(arr6.dtype)
print(arr6.ndim)
```

## 3. Special type numpy arrays

```python
# array of Zeros
arr_zero = np.zeros(5)  #1D
print(arr_zero)
```

```python
arr_zero = np.zeros((2,3))  #2D
print(arr_zero)
```

```python
arr_zero = np.zeros((3,2,3)) #3D
print(arr_zero)
```

```python
# arr of ones
arr_ones = np.ones(5) #1D
print(arr_ones)
```

```python
arr_ones = np.ones((2,3)) #2D
print(arr_ones)
```

```python
arr_ones = np.ones((2,2,4)) #3D
print(arr_ones)
```

```python
# initialize the numpy array with specific value
arr_sep = np.full(5,100)
print(arr_sep)
```

```python
arr_sep = np.full((2,5),100)
print(arr_sep)
```

```python
arr_sep = np.full((2,2,5),100)
print(arr_sep)
```

```python
# identity_matrix is a squared matrix: rows = column
# diagonal values are one and all other zero
id_m = np.eye(4)
print(id_m)
```

# randam package to generate random/ synthetic data 

```python
# rand() function generate random elements betwen 0 and 1
np.random.rand(5) #1D
```

```python
np.random.rand(2,5) #2D
```

```python
np.random.rand(2,2,5)   #3D
```

```python
#  uniform() function generate uniformaly distributed data
# uniformly distributed means all values have equal probability to get selected 
np.random.uniform(low = 1,high =  10, size =5) #1D
```

```python
np.random.uniform(low = 1,high =  10, size =(2,5)) #D
```

```python
# random() function just like rand() function
np.random.random((5))
```

```python
np.random.random((2,5))
```

```python
# randn() : strandard normal distribution function. mean = 0,SD=1
# 99.7% values will be b/w -3 to 3
np.random.randn(5)
```

```python
np.random.randn(2,5)
```

```python
np.random.randn(3,2,5)
```

```python
# normal distributed : here mean & SD can be different from 0 and 1
#loc = mean,scale = SD
np.random.normal(loc = 5, scale=1.2, size=10)
```

```python
np.random.normal(loc = 5, scale=1.2, size=(2,5))
```

```python
# choice() function is  used for categorical data
np.random.seed(42)
np.random.choice(['paradise','fauzi','spirit','Puspa','bahubali'],size=3)
```

```python
np.random.choice(['paradise','fauzi','spirit','Puspa','bahubali'],p=[0.2,0.05,0.05,0.6,0.1],size=10)
```
