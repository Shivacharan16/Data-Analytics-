### 10. Time Series operations

```python
import pandas as pd
date_rng = pd.date_range(start = '2021-01-01', end = '2026-12-31', freq = 'h')
print(date_rng)
```

```python
# pick the 1000 records only
import numpy as np

random_dates = np.random.choice(date_rng, size = 1000, replace = False)
print(len(random_dates))
```

```python
# Creating dataframe using random_dates
canteen_df = pd.DataFrame({
    'timestamp' : pd.to_datetime(random_dates), 
    'order_id' : np.arange(1,1001),  # generate values from 1 to 1000
    'food_category' : np.random.choice(["Samosa", "Egg puff", "Drinks", "Choki Choki", "Ironi_tea"], size = 1000),
    'price' : np.random.randint(10, 500, size = 1000),
    'quantity' : np.random.randint(1, 5, size = 1000)})

canteen_df

# pd.to_datetime : to convert numpy array to pandas date_time format
# np.arange : generate a series of elements e.g 1 , 2, 3, 4, 5...1000
# np.random.choice : used to randomly select categorical data
# np.random.randint : used to generate random integer values
```

```python
# sort timeseries data based upon timestamp
canteen_df = canteen_df.sort_values("timestamp").reset_index(drop=True)

# create new columns with just year ,month and day  
canteen_df['year'] = canteen_df.timestamp.dt.year
canteen_df['month'] = canteen_df.timestamp.dt.month
canteen_df['day'] = canteen_df.timestamp.dt.day
canteen_df['day'] = canteen_df.timestamp.dt.day
canteen_df['hours'] = canteen_df.timestamp.dt.hour
canteen_df['minute'] = canteen_df.timestamp.dt.minute
canteen_df['seconds'] = canteen_df.timestamp.dt.second
canteen_df
```

### 11. Data visualization|

```python
canteen_df.groupby('year')['quantity'].sum().plot(kind = 'pie')
```
